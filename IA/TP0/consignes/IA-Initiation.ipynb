```python
clf = DecisionTreeClassifier()
clf.fit(iris["data"], iris["target"])

plt.figure(figsize=(40,20))  
texts = tree.plot_tree\
(
    clf, 
    filled=True, 
    feature_names = iris["feature_names"], 
    class_names = iris["target_names"], 
    label = "all"
)











#1 / (1 + exp -(w1 * age + w2 * salaire + b1))
#
#w4 *(1 / (1 + exp -(w1 * age + w2 * salaire + b1))) + b2
```

```python
mlp = MLPClassifier(
    hidden_layer_sizes = [2],
    learning_rate= "adaptive", 
    learning_rate_init=0.015
)
mlp = mlp.fit(iris["data"], iris["target"])
print(mlp.coefs_)
```

```python
km = KMeans(n_clusters = 3)
km.fit(iris["data"])
predict = km.predict(iris["data"])
errors = np.count_nonzero(predict!=iris["target"])

print(iris["target"])
print(predict)
print(errors)
```

```python
lr = LinearRegression()
lr.fit(iris["data"], iris["target"])
predict = lr.predict(iris["data"])
errors = np.count_nonzero(predict!=iris["target"])

print(iris["target"])
print(predict)
print(errors)
```
