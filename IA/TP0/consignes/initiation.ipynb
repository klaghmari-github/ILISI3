# import des packages

```python
import sklearn
import matplotlib.pyplot
import pandas
```

# Chargement des données

```python
iris = sklearn.datasets.load_iris()
print(iris["data"], len(iris["data"]), iris["feature_names"])
print(iris["target"], len(iris["target"]), iris["target_names"])
print(iris["filename"])
```

```python
iris['target']
```

```python
df = pandas.DataFrame(data = iris['data'], columns = iris['feature_names'])
df["target"] = iris["target"]
df.shape
```

```python
matplotlib.pyplot.scatter(df["sepal length (cm)"], df["sepal width (cm)"], c = df["target"])
```

```python
df.columns
```

```python
for i in range(len(df.columns) - 2):
    for j in range(i + 1, len(df.columns) - 1):
        matplotlib.pyplot.figure(figsize=(6, 4))
        matplotlib.pyplot.scatter(df[df.columns[i]], df[df.columns[j]], c = df["target"])
        matplotlib.pyplot.xlabel(df.columns[i])
        matplotlib.pyplot.ylabel(df.columns[j])
```

```python
X = df.drop("target", axis = 1)
y = df["target"]
```

```python
clf = sklearn.tree.DecisionTreeClassifier()
clf.fit(X, y)

matplotlib.pyplot.figure(figsize=(40,20))  
texts = sklearn.tree.plot_tree\
(
    clf, 
    filled=True, 
    feature_names = iris["feature_names"], 
    class_names = iris["target_names"], 
    label = "all"
)
```

```python
preds = clf.predict(X)
print(list(preds))
print(list(y))
```

```python
sklearn.metrics.ConfusionMatrixDisplay.from_predictions(
    y,
    preds,
    display_labels = iris["target_names"]
)

matplotlib.pyplot.show()
```
