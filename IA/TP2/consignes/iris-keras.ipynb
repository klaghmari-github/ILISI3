```python
# matplotlib : utile pour la visualisation des données
from matplotlib import pyplot as plt

import numpy as np

# sklearn : bibliothèque de référence pour l'IA en python.
# sklearn.datasets : packages contenant les jeux de données les plus utilisés par la communauté des data scientists.
from sklearn import datasets

import keras

# Sequential est une pile de couches de traitement
from keras.models import Sequential

# Dense est une couche de traitement (neurones)
from keras.layers import Dense

from sklearn.model_selection import train_test_split
```

### Réseau de neurones avec Keras

```python
p(setosa) = 0.4
p(versicolor) = 0.3
p(virginica) = 0.3




6M  2M  500K   200K    50K   20K    5K    2K    500      50     3
```

```python
input_layer = keras.layers.InputLayer(shape = [4,])
hidden_layer = keras.layers.Dense(8, activation = 'sigmoid')
output_layer = keras.layers.Dense(3, activation='softmax')
layers = [input_layer, hidden_layer, output_layer]
model = Sequential(layers)

model.summary()
```

```python
keras.utils.plot_model(
    model,
    to_file="model_plot.png",      # fichier de sortie
    show_shapes=True,              # affiche les shapes
    show_dtype=True,               # affiche les dtypes
    show_layer_names=True,         # affiche les noms des couches
    rankdir="TB",                  # "TB" vertical, "LR" horizontal
    expand_nested=True,            # développe les sous-modèles imbriqués
    dpi=300,                       # meilleure résolution
    show_layer_activations=True,   # affiche les activations si la couche en a une
    show_trainable=True            # affiche si la couche est trainable
)
```
